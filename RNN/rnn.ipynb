```python
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences #RNNs need sequences to have the same length when we train them in batches.
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding ,SimpleRNN ,Dense
```

```python
vacab_size=10000
(x_train,y_train),(x_test,y_test)=imdb.load_data(num_words=vacab_size)
```

```python
print(len(x_train))
print(len(x_test))
```

```python
print("First review length:", len(x_train[0]))
print("second review length:", len(x_train[1]))
print("third review length:", len(x_train[2]))
print("fourth review length:", len(x_train[3]))
```

```python
max_length = 200
X_train = pad_sequences(
    x_train,
    maxlen=max_length,
    padding='post',
    truncating='post'
)
x_test=pad_sequences(
  x_test,maxlen=max_length,
  padding='post',truncating='post'
)
```

```python
model=Sequential([
  Embedding(
    input_dim=vacab_size,
    output_dim=128,
  ),
  SimpleRNN(
    64,
    return_sequences=False
  ),
  Dense(
    1,activation='sigmoid'
  )
])
model.build(input_shape=(None, max_length))
```

```python
model.summary()
```

```python
model.compile(
  optimizer='adam',
  loss='binary_crossentropy',
  metrics=['accuracy']
)
```

```python
history = model.fit(
    X_train,
    y_train,
    epochs=5,
    batch_size=64,
    validation_split=0.2
)
```

```python
test_loss, test_accuracy = model.evaluate(
    x_test,
    y_test
)

print("Test Loss:", test_loss)
print("Test Accuracy:", test_accuracy)
```

```python
y_prob = model.predict(x_test)

y_pred = (y_prob >= 0.5).astype(int)

print("First 10 probabilities:")
print(y_prob[:10])

print("First 10 predictions:")
print(y_pred[:10].flatten())
```
