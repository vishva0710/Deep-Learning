```python
corpus="""hello everyone i am vishva. i intrested to learn AI    
"""
```

```python
print(corpus)
```

```python
# tokenization 
## para -> sentance
from nltk.tokenize import sent_tokenize
sent_tokenize(corpus)
```

```python
sen1=sent_tokenize(corpus)
sen1[0]
```

```python
## para -->words
from nltk.tokenize import word_tokenize
word_tokenize(corpus)
```
