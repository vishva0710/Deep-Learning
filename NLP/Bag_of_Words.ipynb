Bag of Words

```python
#Bag of Words is a technique that converts text into numbers so a machine-learning model can work with the text.
```

```python
#Simple example

#Suppose we have two sentences:

#Sentence 1: "I love Python"
#Sentence 2: "I love AI"
#Step 1 — Create the vocabulary

#Take all unique words:

#I
#love
#Python
#AI

#So our vocabulary is:

#["I", "love", "Python", "AI"]
#Step 2 — Count the words

#For Sentence 1:

#"I love Python"
#Word	I	love	Python	AI
#Sentence 1	1	1	1	0

#So it becomes:

#[1, 1, 1, 0]

#For Sentence 2:

#"I love AI"
#Word	I	love	Python	AI
#Sentence 2	1	1	0	1

#So:

#[1, 1, 0, 1]

#Now the machine can understand the text as numbers.
```

```python
from sklearn.feature_extraction.text import CountVectorizer
sentance=["I love python","i love AI"]
vectorizer=CountVectorizer()
x=vectorizer.fit_transform(sentance)
print(vectorizer.get_feature_names_out())
print(x.toarray())
```
