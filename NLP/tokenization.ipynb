Text preprocessing 

```python
## stemming
# the stemming means it cutting a word down to its basic /root like form by removing prefixes or suffixes
# fro example (palying ,plyed,plays) --> play 
```

```python
words=['eating','eats','eaten','writing','writes','programming','programs','history','finally','finalize']
```

```python
# proterstemmer
from nltk.stem import PorterStemmer
stemming=PorterStemmer()
```

```python
for word in words:
  print(word+'----->'+stemming.stem(word))
```

```python
stemming.stem('congratulations')
```

```python
stemming.stem('sitting')
```

```python
from nltk.stem import RegexpStemmer
stemming=RegexpStemmer('ing$')
print(stemming.stem('playing'))
print(stemming.stem('writing'))
print(stemming.stem('dancing'))
```

```python
#SnowballStemmer 
# is a stemming algorithm in NLTK that removes prefixes/suffixes from words to convert them into a common root-like form.
# It is an improved version of the Porter Stemmer approach and supports multiple languages.
```

```python
from nltk.stem import SnowballStemmer
snowball=SnowballStemmer('english')
for word in words:
  print(word+'----->'+snowball.stem(word))
```

```python
#RegexpStemmer
#     ↓
#You make the rules
#     ↓
#"Remove ing"

#PorterStemmer
#    ↓
#Predefined English stemming rules

#SnowballStemmer
#     ↓
#Improved/flexible stemming framework
#     ↓
#Supports multiple languages
```

```python
stemming.stem('fairly'),stemming.stem('sportingly')
```

```python
snowball.stem('fairly'),snowball.stem('sportingly')
```
