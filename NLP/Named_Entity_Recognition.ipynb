Named Entity Recognition(NER)

```python
#It is an NLP technique used to find important real-world entities (names) inside text and classify them into categories.

#For example:

#"A.P.J. Abdul Kalam was born in Rameswaram and worked at ISRO."

#NER identifies:

#!A.P.J. Abdul Kalam → PERSON
#Rameswaram         → LOCATION
#ISRO               → ORGANIZATION
```

```python
#Common NER categories

#You will commonly see:

#Entity	| Meaning	Example
#PERSON	| Person's name       |	Elon Musk
#ORG	  |Organization	        |Google
#GPE	  | Geopolitical entity |	India
#LOC	  | Location	          |Mount Everest
#DATE	  | Date	              |19 September 2026
#TIME	  | Time	              | 10:30 AM
#MONEY	| Money	              |₹50,000
#PERCENT| Percentage	        |20%
#PRODUCT| Product	            |iPhone
#EVENT  | Event	              | Olympics
```

```python
sentance='''The Eiffel Tower was built from 1887 to 1889 by French engineer
Gustave Eiffel, whose company specialized in building metal frameworks and
structures.'''
```

```python
import nltk
words=nltk.word_tokenize(sentance)
```

```python
tag_element=nltk.pos_tag(words)
```

```python
nltk.ne_chunk(tag_element)
```
