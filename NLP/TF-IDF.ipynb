TF-IDF = Term Frequency × Inverse Document Frequency

```python
#It is a technique used in NLP to convert text into numbers and determine how important a word is in a document.

#The basic idea:

#A word is important if it appears frequently in one document but does NOT appear frequently in many other documents.
```

```python
#TF — Term Frequency

#TF tells us:

#How frequently does a word appear in a particular document?

#Formula:

# TF = \frac{\text{Number of times word appears}}{\text{Total number of words}} $$
#Example

#Document:

#"AI is powerful and AI is useful"

#Total words = 7

#AI appears = 2 times

#So:

# TF(AI)=\frac{2}{7}=0.286 $$

#So the TF of AI = 0.286.
```

```python
#IDF — Inverse Document Frequency
#IDF tells us:

#How rare or unique is this word across all documents?

#Formula:

# IDF = \log\left(\frac{N}{DF}\right) 

#Where:

#N = total number of documents
#DF = number of documents containing the word

#Example:

#Suppose we have 3 documents:

#Document 1:

#AI is powerful

#Document 2:

#AI is useful

#Document 3:

#Python is powerful

#There are 3 documents.

#Let's look at the word AI.

#AI appears in:

#Document 1 
#Document 2 

#So:

# DF(AI)=2 

#Therefore:

#  IDF(AI)=\log\left(\frac{3}{2}\right)

#≈ 0.405
```

```python
#TF-IDF

#Now simply multiply:

# TF-IDF = TF \times IDF 

#So if:

#TF = 0.286
#IDF = 0.405

#Then:

# TF-IDF = 0.286 \times 0.405 

#≈ 0.116

#So the TF-IDF score of AI is approximately 0.116.
```

```python
#But why do we need IDF?

#This is the most important part.

#Imagine these documents:

#Document 1: AI is powerful
#Document 2: AI is useful
#Document 3: AI is interesting

#The word AI appears in every document.

#Therefore, AI doesn't help us distinguish between the documents very much.

#So:

#Common word → Low IDF → Low TF-IDF

#Now imagine:

#Document 1: AI is powerful
#Document 2: AI is useful
#Document 3: Python is interesting

#The word Python appears only in Document 3.

#So Python is more unique to Document 3.

#Rare word → High IDF → High TF-IDF
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
document=[" love Python",
    " love machinelearning",
    " love Python  powerful"]
vector=TfidfVectorizer()
x=vector.fit_transform(document)
print(vector.get_feature_names_out())
print(x.toarray())
```

```python
print(vector.idf_)  
```
