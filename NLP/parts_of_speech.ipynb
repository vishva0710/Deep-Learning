```python
##Parts of Speech = the grammatical role of a word in a sentence.

##For example:

#The boy is running quickly.

#Word	POS	Meaning

#The	    |Determiner	  |specifies the noun
#boy	    |Noun	        |person/thing
#is	      |Verb	        |action/state
#running	|Verb	        |action
#quickly	|Adverb	      |describes how
#.	      |Punctuation	|punctuation
```

```python
#The main POS categories you need for lemmatization are:

#POS	Meaning	Examples

#Noun (n)	       |person/place/thing	          |boy, car, happiness
#Verb (v)	       |action/state	run,            |running, went
#Adjective (a)   |describes a noun      	      |good,  beautiful, bigger
#Adverb (r)	     |describes a verb/adjective	  |quickly, slowly, very
```
